Red Convolucional para MNIST-1D

. El problema principal es construir la red convolucional según las especificaciones.

```python
import torch.nn as nn

model = nn.Sequential(
    # Capa 1: Entrada 1 canal × 40 → 15 canales × 19
    nn.Conv1d(in_channels=1, out_channels=15, kernel_size=3, stride=2, padding=0),
    nn.ReLU(),

    # Capa 2: Entrada 15 canales × 19 → 15 canales × 9
    nn.Conv1d(in_channels=15, out_channels=15, kernel_size=3, stride=2, padding=0),
    nn.ReLU(),

    # Capa 3: Entrada 15 canales × 9 → 15 canales × 4
    nn.Conv1d(in_channels=15, out_channels=15, kernel_size=3, stride=2, padding=0),
    nn.ReLU(),

    # Aplanar: 15 × 4 = 60 características
    nn.Flatten(),

    # Capa lineal final: 60 → 10 clases
    nn.Linear(in_features=60, out_features=10)
)
```

```python
import torch

# Create dummy test data
dummy_input = torch.randn(1, 1, 40) # (batch_size, channels, length)

print(f"Shape of dummy input: {dummy_input.shape}")

# Pass the dummy data through the model
output = model(dummy_input)

print(f"Shape of model output: {output.shape}")
print("Model output (first 5 values):", output[0, :5])
```

```python
import torch, torch.nn as nn

model = nn.Sequential(
    nn.Conv1d(in_channels=1, out_channels=15, kernel_size=3, stride=2, padding=0),
    nn.ReLU(),
    nn.Conv1d(in_channels=15, out_channels=15, kernel_size=3, stride=2, padding=0),
    nn.ReLU(),
    nn.Conv1d(in_channels=15, out_channels=15, kernel_size=3, stride=2, padding=0),
    nn.ReLU(),
    nn.Flatten(),
    nn.Linear(in_features=60, out_features=10)
)

# Prueba con entrada de lote: (1 muestra, 1 canal, 40 pasos)
x_prueba = torch.randn(1, 1, 40)
salida = model(x_prueba)
print(f" Entrada: {x_prueba.shape}")
print(f" Salida:  {salida.shape}")
print(f" Esperada: torch.Size([1, 10])")
print(f" Coincide: {salida.shape == (1, 10)}")
```

Canales de entrada/salida correctos en cada capa

 Kernel=3, paso=2, sin relleno en todas las convoluciones

 ReLU incluida tras cada capa convolucional

 Aplanamiento da exactamente 60 características

 Capa final: 60 → 10 salidas
