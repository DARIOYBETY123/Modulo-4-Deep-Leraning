Regresión Lineal 1D

modelo lineal y la función de pérdida.

```python
# Define 1D linear regression model
def f(x, phi0, phi1):
    # Modelo lineal: ordenada al origen + pendiente × x
    y = phi0 + phi1 * x
    return y
```

PRUEBA

```python
import numpy as np

# Tu función
def f(x, phi0, phi1):
    y = phi0 + phi1 * x
    return y

# Datos del notebook
x = np.array([0.03, 0.19, 0.34, 0.46, 0.78, 0.81, 1.08, 1.18, 1.39, 1.60, 1.65, 1.90])

# Prueba 1: phi0=0.4, phi1=0.2
phi0, phi1 = 0.4, 0.2
y_pred = f(x, phi0, phi1)
print(f" phi0={phi0}, phi1={phi1}")
print(f"   Predicción[0] = {y_pred[0]:.4f} (esperado: 0.4 + 0.2×0.03 = 0.406)")
print(f"   Coincide: {abs(y_pred[0] - 0.406) < 1e-6}")

# Prueba 2: phi0=1.60, phi1=-0.8
phi0, phi1 = 1.60, -0.8
y_pred = f(x, phi0, phi1)
print(f"\n phi0={phi0}, phi1={phi1}")
print(f"   Predicción[0] = {y_pred[0]:.4f} (esperado: 1.6 - 0.8×0.03 = 1.576)")
print(f"   Coincide: {abs(y_pred[0] - 1.576) < 1e-6}")
```

El valor mínimo de pérdida se aproxima con:

```python
import matplotlib.pyplot as plt
import numpy as np

# Define a sample 'y' array, as it's not present in the kernel state.
# This is a placeholder for the true y values you would use for training/evaluation.
# For demonstration, we'll create y values that are somewhat linearly related to x.
y = np.array([0.77, 0.85, 0.95, 1.01, 1.18, 1.2, 1.34, 1.4, 1.5, 1.6, 1.62, 1.7])

# Define the loss function (Mean Squared Error)
def compute_loss(x, y_true, phi0, phi1):
    # Assuming 'f' is defined in a previous cell and is in scope
    y_pred = f(x, phi0, phi1)
    loss = np.mean((y_pred - y_true)**2)
    return loss

# Define the plotting function
def plot(x, y_true, phi0, phi1):
    plt.figure(figsize=(8, 6))
    plt.scatter(x, y_true, label='True data (y)')

    # Generate points for the regression line
    x_plot = np.linspace(x.min(), x.max(), 100)
    y_pred_plot = f(x_plot, phi0, phi1)

    plt.plot(x_plot, y_pred_plot, color='red', label=f'Regression line: y = {phi0:.2f} + {phi1:.2f}x')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Linear Regression Model')
    plt.legend()
    plt.grid(True)
    plt.show()

phi0 = 0.75   # intercepto
phi1 = 0.55   # pendiente

plot(x, y, phi0, phi1)
print(f'Tu pérdida = {compute_loss(x, y, phi0, phi1):3.2f}')
#  Pérdida mínima ≈ 0.36
```
