construye una red neuronal para clasificación binaria (salida: 0 o 1) y usa la distribución de Bernoulli para calcular la verosimilitud.

```python
import numpy as np
```

```python
def sigmoid(model_out):
    model_out = np.squeeze(model_out)
    sig_model_out = 1 / (1 + np.exp(-model_out))
    return sig_model_out
```

```python
# Pruebas rápidas
print("sigmoid(0)   =", sigmoid(0))    # Debe ser ≈ 0.5
print("sigmoid(2)   =", sigmoid(2))    # Debe ser ≈ 0.8808
print("sigmoid(-2)  =", sigmoid(-2))   # Debe ser ≈ 0.1192
print("sigmoid(100) =", sigmoid(100))  # Debe ser ≈ 1.0
print("sigmoid(-100)=", sigmoid(-100)) # Debe ser ≈ 0.0
```

Distribución de Bernoulli

Probabilidad de la clase observada y dado el parámetro λ

```python
def bernoulli_distribution(y, lambda_param):
    lambda_param = np.squeeze(lambda_param)
    y = np.squeeze(y)
    # Si y=1 → λ ; si y=0 → 1-λ
    prob = np.where(y == 1, lambda_param, 1 - lambda_param)
    return prob
```

```python
# Pruebas oficiales del cuaderno
print("Respuesta correcta = 0.800, Tu respuesta = %3.3f" % bernoulli_distribution(0, 0.2))
print("Respuesta correcta = 0.200, Tu respuesta = %3.3f" % bernoulli_distribution(1, 0.2))
```

Likelihood (Verosimilitud)

Producto de las probabilidades de cada dato

```python
def compute_likelihood(y_train, lambda_param):
    lambda_param = np.squeeze(lambda_param)
    prob_punto = bernoulli_distribution(y_train, lambda_param)
    likelihood = np.prod(prob_punto)
    return likelihood
```

```python
# Definición de datos de ejemplo para x_train y y_train
x_train = np.array([[0.5], [0.2], [0.8], [0.1]])
y_train = np.array([1, 0, 1, 0])
```

```python
def get_parameters():
    # Estos son valores de ejemplo, ajusta según la arquitectura de tu red
    beta_0 = 0.5
    omega_0 = np.array([[-0.5]]) # Ejemplo de un peso para la primera capa
    beta_1 = 0.2
    omega_1 = np.array([[0.7]]) # Ejemplo de un peso para la segunda capa
    return beta_0, omega_0, beta_1, omega_1
```

```python
def shallow_nn(x, beta_0, omega_0, beta_1, omega_1):
    # Implementación de una red neuronal poco profunda (shallow neural network)
    # Esto es una simplificación, la arquitectura real dependerá de tu problema.

    # Primera capa (oculta)
    hidden_layer_input = np.dot(x, omega_0) + beta_0
    hidden_layer_output = hidden_layer_input # Puedes aplicar una función de activación aquí (e.g., ReLU)

    # Capa de salida
    output = np.dot(hidden_layer_output, omega_1) + beta_1
    return output
```

```python
# Prueba oficial
beta_0, omega_0, beta_1, omega_1 = get_parameters()
model_out = shallow_nn(x_train, beta_0, omega_0, beta_1, omega_1)
lambda_train = sigmoid(model_out)
likelihood = compute_likelihood(y_train, lambda_train)

print("Respuesta correcta = 0.000070237")
print(f"Tu respuesta      = {likelihood:.9f}")
```

Log-Likelihood Negativa

Menos la suma de los logaritmos → numéricamente estable

```python
def compute_negative_log_likelihood(y_train, lambda_param):
    lambda_param = np.squeeze(lambda_param)
    log_prob = np.log(bernoulli_distribution(y_train, lambda_param))
    nll = -np.sum(log_prob)
    return nll
```

```python
# Prueba oficial
beta_0, omega_0, beta_1, omega_1 = get_parameters()
model_out = shallow_nn(x_train, beta_0, omega_0, beta_1, omega_1)
lambda_train = sigmoid(model_out)
nll = compute_negative_log_likelihood(y_train, lambda_train)

print("Respuesta correcta = 9.563639387")
print(f"Tu respuesta      = {nll:.9f}")
```
