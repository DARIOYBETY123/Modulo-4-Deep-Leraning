Función: Gradientes Normalizados

```python
def normalized_gradients(start_posn, n_steps, alpha, epsilon=1e-20):
    grad_path = np.zeros((2, n_steps+1))
    grad_path[:,0] = start_posn[:,0]

    for c_step in range(n_steps):
        # Medir el gradiente m = g_t
        m = get_loss_gradient(grad_path[0,c_step], grad_path[1,c_step])

        #  TODO RESUELTO: Calcular el gradiente cuadrado v = g_t^2
        v = m ** 2

        #  TODO RESUELTO: Actualización normalizada
        grad_path[:,c_step+1] = grad_path[:,c_step] - alpha * m / (np.sqrt(v) + epsilon)

    return grad_path
```

```python
def adam(start_posn, n_steps, alpha, beta=0.9, gamma=0.99, epsilon=1e-20):
    grad_path = np.zeros((2, n_steps+1))
    grad_path[:,0] = start_posn[:,0]

    m = np.zeros_like(grad_path[:,0])  # Promedio móvil del gradiente
    v = np.zeros_like(grad_path[:,0])  # Promedio móvil del gradiente cuadrado

    for c_step in range(n_steps):
        t = c_step + 1  # Contador empieza en 1

        # Medir el gradiente
        grad = get_loss_gradient(grad_path[0,c_step], grad_path[1,c_step])

        #  TODO RESUELTO: Actualizar promedio móvil del gradiente
        m = beta * m + (1 - beta) * grad

        #  TODO RESUELTO: Actualizar promedio móvil del gradiente cuadrado
        v = gamma * v + (1 - gamma) * (grad ** 2)

        #  TODO RESUELTO: Corrección de sesgo
        m_tilde = m / (1 - np.power(beta, t))
        v_tilde = v / (1 - np.power(gamma, t))

        #  TODO RESUELTO: Actualización de Adam
        grad_path[:,c_step+1] = grad_path[:,c_step] - alpha * m_tilde / (np.sqrt(v_tilde) + epsilon)

    return grad_path
```
