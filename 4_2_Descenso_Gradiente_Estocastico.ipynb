Descenso por Gradiente Estocástico con Mini-Batches

```python
def stochastic_gradient_descent_step(phi, data, alpha, batch_size):
    n_total = data.shape[1]  # Número total de datos

    # Generar índices aleatorios sin repetición
    indices = np.random.permutation(n_total)

    # Seleccionar los primeros 'batch_size' índices
    batch_idx = indices[:batch_size]

    # Extraer solo los datos del mini-batch
    data_x_batch = data[0, batch_idx]
    data_y_batch = data[1, batch_idx]

    # Calcular el gradiente SÓLO con el mini-batch
    grad = compute_gradient(data_x_batch, data_y_batch, phi)

    # Actualizar parámetros
    phi = phi - alpha * grad

    return phi
```

Tasa de Aprendizaje con Reducción Programada

```python
# Para implementar el scheduler, agregamos un contador de pasos externo
def stochastic_gradient_descent_step_with_scheduler(phi, data, alpha, batch_size, step, M=10, beta=0.5):
    # Reducir alpha por factor beta cada M iteraciones
    alpha_current = alpha * (beta ** (step // M))

    n_total = data.shape[1]
    indices = np.random.permutation(n_total)
    batch_idx = indices[:batch_size]
    grad = compute_gradient(data[0, batch_idx], data[1, batch_idx], phi)
    phi = phi - alpha_current * grad
    return phi
```

Preguntas de experimentación

¿α demasiado grande? → Los parámetros "saltan" demasiado → la pérdida crece en lugar de disminuir → diverge.

¿α demasiado pequeño? → Avance minúsculo → necesita cientos de pasos para converger.

¿Batch size pequeño? → Más ruido en la trayectoria, pero avanza más rápido por época.

¿Punto inicial distinto? → La superficie tiene varios valles → puede quedarse atascado en un mínimo local.
