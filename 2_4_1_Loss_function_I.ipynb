2.4.1 Loss function I

 Función de Distribución Normal

```python
def normal_distribution(y, mu, sigma):
    # Exponente
    exponente = - (y - mu) ** 2 / (2 * sigma ** 2)
    # Parte de normalización
    denominador = sigma * np.sqrt(2 * math.pi)
    # Probabilidad final
    prob = np.exp(exponente) / denominador
    return prob
```

Likelihood (Probabilidad Conjunta)

Es el producto de las probabilidades de cada dato

```python
def compute_likelihood(y_train, mu, sigma):
    # Asegurar que tengan la misma forma
    mu = np.squeeze(mu)
    # Probabilidad de cada punto
    prob_punto = normal_distribution(y_train, mu, sigma)
    # Producto de todas las probabilidades
    likelihood = np.prod(prob_punto)
    return likelihood
```

Código Completo de Prueba

```python
import numpy as np
import math

# 1. Distribución normal ✅ COMPLETA
def normal_distribution(y, mu, sigma):
    mu = np.squeeze(mu)
    y = np.squeeze(y)
    exponente = - (y - mu) ** 2 / (2 * sigma ** 2)
    denominador = sigma * np.sqrt(2 * np.pi)
    prob = np.exp(exponente) / denominador
    return prob

# 2. Tu Likelihood ✅
def compute_likelihood(y_train, mu, sigma):
    mu = np.squeeze(mu)
    y_train = np.squeeze(y_train)
    prob_punto = normal_distribution(y_train, mu, sigma)
    likelihood = np.prod(prob_punto)
    return likelihood

# === DATOS Y PARÁMETROS ===
def ReLU(preactivation):
    return preactivation.clip(0.0)

def shallow_nn(x, beta_0, omega_0, beta_1, omega_1):
    n_data = x.size
    x = np.reshape(x,(1,n_data))
    h1 = ReLU(np.matmul(beta_0,np.ones((1,n_data))) + np.matmul(omega_0,x))
    y = np.matmul(beta_1,np.ones((1,n_data))) + np.matmul(omega_1,h1)
    return y

def get_parameters():
    beta_0 = np.zeros((3,1))
    omega_0 = np.zeros((3,1))
    beta_1 = np.zeros((1,1))
    omega_1 = np.zeros((1,3))
    beta_0[0,0]=0.3; beta_0[1,0]=-1.0; beta_0[2,0]=-0.5
    omega_0[0,0]=-1.0; omega_0[1,0]=1.8; omega_0[2,0]=0.65
    beta_1[0,0]=0.1
    omega_1[0,0]=-2.0; omega_1[0,1]=-1.0; omega_1[0,2]=7.0
    return beta_0, omega_0, beta_1, omega_1

x_train = np.array([0.09291784,0.46809093,0.93089486,0.67612654,0.73441752,0.86847339,
                   0.49873225,0.51083168,0.18343972,0.99380898,0.27840809,0.38028817,
                   0.12055708,0.56715537,0.92005746,0.77072270,0.85278176,0.05315950,
                   0.87168699,0.58858043])
y_train = np.array([-0.25934537,0.18195445,0.651270150,0.13921448,0.09366691,0.30567674,
                    0.372291170,0.20716968,-0.08131792,0.51187806,0.16943738,0.3994327,
                    0.019062570,0.55820410,0.452564960,-0.1183121,0.02957665,-1.24354444,
                    0.248038840,0.26824970])

# === EJECUTAR PRUEBA ===
beta_0, omega_0, beta_1, omega_1 = get_parameters()
mu_pred = shallow_nn(x_train, beta_0, omega_0, beta_1, omega_1)
sigma = 0.2

likelihood = compute_likelihood(y_train, mu_pred, sigma)
print(f"✅ Esperado: 0.000010624")
print(f"✅ Obtenido: {likelihood:.9f}")
```

Log-Likelihood Negativa (NLL)

En lugar de multiplicar, sumamos logaritmos y cambiamos el signo

```python
def compute_negative_log_likelihood(y_train, mu, sigma):
    mu = np.squeeze(mu)
    # Logaritmo de la probabilidad de cada punto
    log_prob = np.log(normal_distribution(y_train, mu, sigma))
    # Menos la suma → minimizar esto = maximizar likelihood
    nll = -np.sum(log_prob)
    return nll
```

```python
# Prueba la función
beta_0, omega_0, beta_1, omega_1 = get_parameters()

# Usa la red neuronal para predecir la media
mu_pred = shallow_nn(x_train, beta_0, omega_0, beta_1, omega_1)

# Establece la desviación estándar
sigma = 0.2

# Calcula la log-likelihood negativa
nll = compute_negative_log_likelihood(y_train, mu_pred, sigma)

# ✅ Verifica el resultado
print("Respuesta correcta = 11.452419564")
print(f"Tu respuesta      = {nll:.9f}")
```

Suma de Cuadrados

```python
def compute_sum_of_squares(y_train, y_pred):
    y_pred = np.squeeze(y_pred)
    sum_of_squares = np.sum((y_train - y_pred) ** 2)
    return sum_of_squares
```

```python
# Prueba la función
beta_0, omega_0, beta_1, omega_1 = get_parameters()

# Predicción de la red neuronal
y_pred = shallow_nn(x_train, beta_0, omega_0, beta_1, omega_1)

# Calcula la suma de cuadrados
sum_of_squares = compute_sum_of_squares(y_train, y_pred)

# ✅ Verifica el resultado
print("Respuesta correcta = 2.020992572")
print(f"Tu respuesta      = {sum_of_squares:.9f}")
```

Maximizar likelihood ≡ Minimizar NLL ≡ Minimizar suma de cuadrados. Los tres dan el mismo valor óptimo de beta_1.

Al variar beta_1 (desplazamiento vertical):
La likelihood tiene un pico (máximo)
La NLL tiene un valle (mínimo)
La suma de cuadrados también tiene su mínimo
Los tres coinciden en el mismo valor ≈ 0.27

Al variar sigma (ruido/incertidumbre):
La suma de cuadrados NO cambia → no depende de σ
La likelihood y NLL SÍ cambian → hay un valor óptimo de σ ≈ 0.20–0.22
