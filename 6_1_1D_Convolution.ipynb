Convolución 1D

Convolución: Kernel 3, Paso 1, Dilatación 1, Relleno de ceros

```python
def conv_3_1_1_zp(x_in, omega):
    n = len(x_in)
    x_out = np.zeros(n)

    for i in range(n):
        # Índices con relleno de ceros
        x_m1 = x_in[i-1] if i-1 >= 0 else 0.0
        x_0  = x_in[i]
        x_p1 = x_in[i+1] if i+1 < n else 0.0

        # Suma ponderada
        x_out[i] = omega[0] * x_m1 + omega[1] * x_0 + omega[2] * x_p1

    return x_out
```

```python
def conv_3_2_1_zp(x_in, omega):
    n_in = len(x_in)
    n_out = int(np.ceil(n_in / 2))
    x_out = np.zeros(n_out)

    for j in range(n_out):
        i = j * 2  # posición central en entrada
        x_m1 = x_in[i-1] if i-1 >= 0 else 0.0
        x_0  = x_in[i]
        x_p1 = x_in[i+1] if i+1 < n_in else 0.0

        x_out[j] = omega[0] * x_m1 + omega[1] * x_0 + omega[2] * x_p1

    return x_out
```

Convolución: Kernel 5, Paso 1, Dilatación 1

```python
def conv_5_1_1_zp(x_in, omega):
    n = len(x_in)
    x_out = np.zeros(n)

    for i in range(n):
        x_m2 = x_in[i-2] if i-2 >= 0 else 0.0
        x_m1 = x_in[i-1] if i-1 >= 0 else 0.0
        x_0  = x_in[i]
        x_p1 = x_in[i+1] if i+1 < n else 0.0
        x_p2 = x_in[i+2] if i+2 < n else 0.0

        x_out[i] = omega[0]*x_m2 + omega[1]*x_m1 + omega[2]*x_0 + omega[3]*x_p1 + omega[4]*x_p2

    return x_out
```

Convolución: Kernel 3, Paso 1, Dilatación 2

```python
def conv_3_1_2_zp(x_in, omega):
    n = len(x_in)
    x_out = np.zeros(n)

    for i in range(n):
        d = 2  # dilatación
        x_md = x_in[i-d] if i-d >= 0 else 0.0
        x_0  = x_in[i]
        x_pd = x_in[i+d] if i+d < n else 0.0

        x_out[i] = omega[0] * x_md + omega[1] * x_0 + omega[2] * x_pd

    return x_out
```

Matriz de Convolución (Kernel 3, Paso 1, Dilatación 1)

```python
def get_conv_mat_3_1_1_zp(n_out, omega):
    omega_mat = np.zeros((n_out, n_out))

    for i in range(n_out):
        if i-1 >= 0:
            omega_mat[i, i-1] = omega[0]
        omega_mat[i, i] = omega[1]
        if i+1 < n_out:
            omega_mat[i, i+1] = omega[2]

    return omega_mat
```

PRUEBA

```python
import numpy as np

# Señal del notebook
x = np.array([5.2, 5.3, 5.4, 5.1, 10.1, 10.3, 9.9, 10.3, 3.2, 3.4, 3.3, 3.1])

# Tu función de convolución directa
def conv_3_1_1_zp(x_in, omega):
    n = len(x_in)
    x_out = np.zeros(n)
    for i in range(n):
        x_m1 = x_in[i-1] if i-1 >= 0 else 0.0
        x_0  = x_in[i]
        x_p1 = x_in[i+1] if i+1 < n else 0.0
        x_out[i] = omega[0]*x_m1 + omega[1]*x_0 + omega[2]*x_p1
    return x_out

# Tu función de matriz de convolución
def get_conv_mat_3_1_1_zp(n_out, omega):
    omega_mat = np.zeros((n_out, n_out))

    for i in range(n_out):
        if i-1 >= 0:
            omega_mat[i, i-1] = omega[0]
        omega_mat[i, i] = omega[1]
        if i+1 < n_out:
            omega_mat[i, i+1] = omega[2]

    return omega_mat

# ==============================================
# PRUEBA: comparar convolución directa vs matricial
# ==============================================
omega = np.array([-1.0, 0.5, -0.2])

# Método 1: convolución directa
h6 = conv_3_1_1_zp(x, omega)

# Método 2: producto matricial
omega_mat = get_conv_mat_3_1_1_zp(len(x), omega)
h7 = np.matmul(omega_mat, x)

print(" Convolución directa h6:")
print(np.round(h6, 4))

print("\n Producto matricial h7:")
print(np.round(h7, 4))

print("\n ¿Son idénticos?")
coinciden = np.allclose(h6, h7)
print(f"   {' SÍ — Resultados idénticos!' if coinciden else '❌ NO coinciden'}")
```

Pregunta Teórica Final
¿Qué ocurre si aplicamos dos veces la convolución? ¿Se puede expresar como una sola?
✅ Respuesta:
Aplicar dos convoluciones sucesivas = una sola convolución cuyo kernel es la convolución de ambos kernels originales (propiedad de composición).

Es decir: componer dos convoluciones = convolver sus kernels.
